```python
import pandas as pd  # Pandas sirve para crear y manipular bases de datos 
import matplotlib.pyplot as plt #este paquete sirve para hacer gràficas
import numpy as np #
path = 'C:\\Users\\Personal\\OneDrive - ESCUELA SUPERIOR POLITECNICA DE CHIMBORAZO\\Documentos\\PYTHON'

documento = f'{path}\\Libro1.csv'     # f se coloca antes de las comillas para poder ingresar variables dentro del texto, estas variables deben ir entre corchetes. 

base_datos = pd.read_csv(documento)    # Nos ayuda a leer la base de datos en un DataFrame 

print(base_datos)
print (base_datos.describe())
#para crear una base de datos
base_datos['Irradiance_2']= base_datos['Irradiance']*10000 #->W/cm2, Irradiance_2 es la nueva columna a crear, Irradiance es la columna que vamos a agarrar para acerle un cambio

print(base_datos)
plt.plot(base_datos['Time (seg)'], base_datos['Irradiance_2']) #para llamar a la columna y usarla en este caso como grafiaca
plt.show()

def funcion(S):  # 'S' valor que entra
    min=S/60     #calculos
    return  min # nombre que de la var que queremos que devuelva

base_datos['Minutos']=base_datos['Time (seg)'].apply(funcion) 

print(base_datos)

def power(x,min):
    E=x*min
    return E

base_datos ['Energy']=power(base_datos['Power (W)'],base_datos['Minutos']) #para trabajar con dos variables desde la funcion
print(base_datos)
```

```python
import matplotlib.pyplot as plt
import pandas as pd  # Pandas sirve para crear y manipular bases de datos 
import seaborn as sb

path = 'C:\\Users\\Personal\\OneDrive - ESCUELA SUPERIOR POLITECNICA DE CHIMBORAZO\\Documentos\\PYTHON'
documento = f'{path}\\semillasbd.csv'     # f se coloca antes de las comillas para poder ingresar variables dentro del texto, estas variables deben ir entre corchetes. 
#BdeDAtos-semilllas
BD = pd.read_csv(documento, encoding='latin1')    # Nos ayuda a leer la base de datos en un DataFrame 
print(BD)
print(BD.columns) #pa saber el nombre de la base de datos
#print(DB['Diám.ecuatorial'])
sb.histplot(data=BD['Diám.ecuatorial'], kde=True,bins=15)
plt.savefig(f'{path}\\Histograma_ecuatorial.png', dpi=600)
plt.show()
sb.histplot(data=BD['Diám.Polar'], kde=True,bins=15)
plt.savefig(f'{path}\\Histograma_polar.png', dpi=600)
plt.show()
sb.kdeplot(data=BD['Diám.ecuatorial'], fill=True, color='violet')#
plt.savefig(f'{path}\\curva_histrograma.png', dpi=600)
plt.show()
sb.scatterplot(data=BD['Diám.ecuatorial'])#ver la dispersion de los datos, ver los datos de abajo y arriba para determinar si hay dispersion o no
plt.savefig(f'{path}\\grafica_dispersion.png', dpi=600)
plt.show()
sb.boxplot(data=(BD['Diám.ecuatorial'],BD['Diám.Polar']))# una grafica de una caja donde su anchura y largo representan datos estadisticos, circulos son datos atipicos->serán datos a discriminar / tambien si +5 desv. estan. puede ser que nada tiene que ver, por ejemplo puede ser otro tipo de semilla, un nuevo descubrimiento
plt.savefig(f'{path}\\grafica_caja.png', dpi=600)
plt.show()
sb.violinplot(data=(BD['Diám.ecuatorial'],BD['Diám.Polar']))#distribuye datos, describe, pero no da datos atípicos solo se los puede observar con las colitas, entre más delgaditas sean
plt.savefig(f'{path}\\grafica_violin.png', dpi=600)
plt.show()
#correlacion lineal 
#correlacion de pearson -1=<P(x)=<1, mientras mas se acerque a -1 valores solo entre 0-1
sb.countplot(data=(BD['Diám.ecuatorial'],BD['Diám.Polar']))#grafica para contar
plt.savefig(f'{path}\\contador.png', dpi=600)
plt.show()

sb.jointplot(data=(BD['Diám.ecuatorial'],BD['Diám.Polar'])) #arriba distrib. total de los datos, derecha distrib. individual todo en correlacion a conteos , en rago de tamaños...en la gráfica se puede obseervar que los diametros polares siempre estan siendo más grandes que las d.ecuat.
plt.savefig(f'{path}\\correlaciones_3tipos.png', dpi=600)
plt.show()
#matrix que da las correlaciones entre las variables
corr_matrix=BD.corr()
plt.figure(figsize=(7,6))
plt.imshow(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlación")
plt.savefig(f'{path}\\correlacion_matrix.png', dpi=600)
plt.show()
#si tira pal azul es inver. proporcional y pal rojo directamente propor.
#Regresion lineal solo (sb.regplot(y=BD['Diám.ecuatorial'],x=BD['Diám.Polar']))
#regresion cuadratica añadir order=4
sb.regplot(y=BD['Diám.ecuatorial'],x=BD['Diám.Polar'],order=4)
plt.savefig(f'{path}\\regresion_lineal.png', dpi=600)
plt.show()
```
